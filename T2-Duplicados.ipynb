```python
import pandas as pd
import numpy as np
import seaborn as sns

df = df = pd.read_csv("titanic_limpio.csv")
```

```python
df.duplicated().sum()
```

Cantidad de duplicados

```python
df[df.duplicated()]
```

Ver filas duplicadas

```python
df[df.duplicated(keep=False)]
```

Dos pasajeros pueden compartir muchos datos y eso no significa necesariamente que sean duplicados reales, así que conviene comprobamos antes de borrar lo que pandas considera duplicado

```python
duplicados = df[df.duplicated(keep=False)].sort_values(by=list(df.columns))
duplicados
```

duplicated() compara todas las columnas. Si dos filas son exactamente iguales, pandas las considera duplicadas, pero eso no demuestra por sí solo que sean un error del dataset

```python
df = df.drop_duplicates()
```

Una vez comprobados eliminamos duplicados

```python
df.duplicated().sum()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns

df = df = pd.read_csv("titanic_limpio.csv")
```

```python
porcentaje_supervivientes = df["survived"].mean() * 100

print(f"Supervivientes: {porcentaje_supervivientes:.2f}%")
```

```python
supervivencia_sexo = df.groupby("sex")["survived"].mean() * 100

print(supervivencia_sexo)
```

```python
supervivencia_clase = df.groupby("pclass")["survived"].mean() * 100

print(supervivencia_clase)
```

```python
edad_supervivencia = df.groupby("survived")["age"].mean()

print(edad_supervivencia)
```

```python
supervivencia_edad = df.groupby("franja_edad")["survived"].mean() * 100

print(supervivencia_edad)
```

```python
sexo_clase = df.groupby(["sex", "pclass"])["survived"].mean() * 100

print(sexo_clase)
```

```python
df.to_csv("titanic_limpio.csv", index=False)
```

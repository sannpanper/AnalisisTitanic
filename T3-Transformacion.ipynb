```python
import pandas as pd
import numpy as np
import seaborn as sns

df = pd.read_csv("titanic_limpio.csv")
```

```python
df["franja_edad"] = np.select(
    [
        df["age"] < 18,
        df["age"] < 60,
        df["age"] >= 60
    ],
    [
        "Menor",
        "Adulto",
        "Mayor"
    ],
    default="Desconocido"
)
```

```python
df[["age", "franja_edad"]].head(20)
```

```python
df["franja_edad"].value_counts()
```

```python
df["franja_edad"].value_counts(normalize=True) * 100
```

```python
df["familiares"] = df["sibsp"] + df["parch"]
```

Se ha creado la columna franja_edad mediante la función np.select() de NumPy. Esta nueva variable permite clasificar a los pasajeros en tres grupos según su edad: menores de 18 años, adultos entre 18 y 59 años y mayores de 60 años.

También se ha creado la columna familiares, calculada a partir de la suma de las columnas sibsp y parch. Esta variable representa el número de familiares que viajaban con cada pasajero según los datos disponibles.

Estas transformaciones permiten trabajar posteriormente con grupos de edad y tamaño de familia durante el análisis exploratorio.

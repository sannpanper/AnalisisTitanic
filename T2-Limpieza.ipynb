T2- Valores nulos (Maria)

```python
import pandas as pd
import numpy as np
import seaborn as sns

df = sns.load_dataset("titanic")
```

Carga del dataset

```python
df.isna().sum()
```

Detectar valores nulos

```python
(df.isna().sum() / len(df)) * 100
```

Util. Permite saber qué columnas tienen problemas importantes

```python
df["age"] = df["age"].fillna(df["age"].median())
```

Limpieza de edad. La edad tiene algunos valores extremos y la mediana es mejor para ellos. Además, es una variable numérica

```python
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
```

embarked es una variable categórica. No podemos ponerle una media. La moda es el valor que aparece más veces

```python
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
```

embark_town tiene 2 valores nulos y es también categórica.

```python
df = df.drop(columns=["deck"])
```

La columna deck presenta 688 valores nulos de un total de 891 registros, aproximadamente el 77,2 %. Debido a que la mayoría de los datos de esta variable están ausentes, se ha decidido eliminar la columna en lugar de imputar sus valores, ya que una imputación tan extensa introduce información artificial.

```python
df.isna().sum()
```

Comprobación de limpieza

```python
df.to_csv("titanic_limpio.csv", index=False)
```

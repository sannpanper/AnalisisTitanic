```python
import pandas as pd

df = pd.read_csv("titanic_limpio.csv")
```

```python
df.head()
```

```python
porcentaje_supervivientes = df["survived"].mean() * 100

print(f"Porcentaje de supervivientes: {porcentaje_supervivientes:.2f}%")
```

```python
supervivencia_sexo = df.groupby("sex")["survived"].mean() * 100

print(supervivencia_sexo)
```

```python
supervivencia_clase = df.groupby("pclass")["survived"].mean() * 100

print(supervivencia_clase)
```

```python
edad_supervivencia = df.groupby("survived")["age"].mean()

print(edad_supervivencia)
```

0= fallecido
1=superviviente

```python
supervivencia_edad = df.groupby("franja_edad")["survived"].mean() * 100

print(supervivencia_edad)
```

```python
sexo_clase = df.groupby(["sex", "pclass"])["survived"].mean() * 100

print(sexo_clase)
```

Varios filtros a continuacion:


```python
menores = df[df["age"] < 18]

print(menores)
```

```python
print("Número de menores:", len(menores))
```

```python
mujeres = df[df["sex"] == "female"]

print("Número de mujeres:", len(mujeres))
```

Se han realizado diferentes agrupaciones, filtros y cálculos estadísticos para estudiar la supervivencia según el sexo, la clase y la edad de los pasajeros.
